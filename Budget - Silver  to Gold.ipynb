```python
df = spark.read.table("workspace.joint_account_budget_silver.joint_budget")
display(df)
```

```python
from pyspark.sql.functions import sum, year, month
gold_df = df.groupBy(
        year(df["date"]).alias("year"), 
        month(df["date"]).alias("month"),
        "category",
    ).agg(
        sum("amount").alias("budget")
    )
display(gold_df)
```
