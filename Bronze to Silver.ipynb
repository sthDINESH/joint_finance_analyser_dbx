# Joint statements - Bronze to Silver transformations

```python
df = spark.table("workspace.joint_statements_bronze.joint_statements_google_drive")
# df.show(10)
```

```python
df = df.drop("Sort Code", "Account Number", "_gdrive_metadata", "_file_metadata")
df = df.withColumnsRenamed({"Transaction Date": "date", "Transaction Type": "type", "Transaction Description": "description", "Debit Amount": "debit", "Credit Amount": "credit", "Balance": "balance"})
# df.show(10)
```

```python
from pyspark.sql.functions import to_date

df = df.withColumn("date", to_date(df["date"], "dd/MM/yyyy")) \
    .withColumn("debit", df["debit"].cast("double")) \
    .withColumn("credit", df["credit"].cast("double")) \
    .withColumn("balance", df["balance"].cast("double"))
```

```python
df.write.format("delta").mode("overwrite").saveAsTable("workspace.joint_statements_silver.joint_statements")
print("Saved to workspace.joint_statements_silver.joint_statements")

# display(spark.table("workspace.joint_statements_silver.joint_statements").limit(10))
```
